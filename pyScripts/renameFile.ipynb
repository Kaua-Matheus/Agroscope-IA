```python
import os
import shutil
```

```python
PATH = "/media/kaua-matheus/HD320/Data/Agroscope/Generalista_Amostra_Maior"
```

```python
def preview_reorganizacao(path_base):
    """
    Mostra o que seria feito sem executar as ações
    """
    culturas = ['Wheat', 'Soybean', 'Corn']
    
    for cultura in culturas:
        cultura_path = os.path.join(path_base, cultura)
        
        if not os.path.exists(cultura_path):
            print(f"❌ Pasta {cultura} não encontrada!")
            continue
            
        print(f"\n📁 {cultura}")
        print("-" * 50)
        
        contador = 1
        total_arquivos = 0
        
        for item in os.listdir(cultura_path):
            subpasta_path = os.path.join(cultura_path, item)
            
            if os.path.isdir(subpasta_path):
                arquivos_subpasta = []
                
                for arquivo in os.listdir(subpasta_path):
                    arquivo_path = os.path.join(subpasta_path, arquivo)
                    
                    if os.path.isfile(arquivo_path):
                        _, extensao = os.path.splitext(arquivo)
                        novo_nome = f"{cultura}_{contador:04d}{extensao}"
                        arquivos_subpasta.append(f"{arquivo} -> {novo_nome}")
                        contador += 1
                        total_arquivos += 1
                
                if arquivos_subpasta:
                    print(f"  📂 {item} ({len(arquivos_subpasta)} arquivos)")
                    for mudanca in arquivos_subpasta[:3]:  # Mostrar primeiros 3
                        print(f"    {mudanca}")
                    if len(arquivos_subpasta) > 3:
                        print(f"    ... e mais {len(arquivos_subpasta) - 3} arquivos")
        
        print(f"  📊 Total: {total_arquivos} arquivos")

# Executar preview primeiro
preview_reorganizacao(PATH)
```
