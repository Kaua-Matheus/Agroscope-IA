```python
import cv2

import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("img_test.jpg", cv2.IMREAD_GRAYSCALE)
```

## Aplicando padrões default

```python
laplacian = cv2.Laplacian(img, cv2.CV_64F, ksize=5)
scharrx = cv2.Scharr(img, cv2.CV_64F, 1, 0)
scharry = cv2.Scharr(img, cv2.CV_64F, 1, 0)
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=5)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=5)
```

```python
plt.subplot(2, 2, 1), plt.imshow(img, cmap='gray')
plt.title("Imagem Original")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
plt.subplot(1, 1, 1), plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
plt.subplot(1, 1, 1), plt.imshow(scharrx, cmap='gray')
plt.title("ScharrX")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
plt.subplot(1, 1, 1), plt.imshow(scharry, cmap='gray')
plt.title("ScharrY")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
plt.subplot(1, 1, 1), plt.imshow(sobelx, cmap='gray')
plt.title("SobelX")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
plt.subplot(1, 1, 1), plt.imshow(sobely, cmap='gray')
plt.title("SobelY")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

## Aplicando Blur no Scharr

```python
blur = cv2.GaussianBlur(img, (3, 3), 0)
```

```python
grad_x = cv2.Scharr(blur, cv2.CV_64F, 1, 0)
grad_y = cv2.Scharr(blur, cv2.CV_64F, 0, 1)
magnitude = cv2.magnitude(grad_x, grad_y)
```

```python
plt.subplot(1, 1, 1), plt.imshow(grad_x, cmap='gray')
plt.title("Grad X")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
plt.subplot(1, 1, 1), plt.imshow(magnitude, cmap='gray')
plt.title("Grad Magnitude")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
sobel_x = cv2.Scharr(sobelx, cv2.CV_64F, 1, 0)
sobel_y = cv2.Scharr(sobely, cv2.CV_64F, 0, 1)
sobel_magnitude = cv2.magnitude(sobel_x, sobel_y)
```

```python
plt.subplot(1, 1, 1), plt.imshow(sobel_magnitude, cmap='gray')
plt.title("Sobel Magnitude")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```

### Aplicando Normalize

```python
scharr_edges = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX)
scharr_edges = scharr_edges.astype(np.uint8)
```

```python
plt.subplot(1, 1, 1), plt.imshow(scharr_edges, cmap='gray')
plt.title("Grad Edges")

# Zera os eixos
plt.xticks([]), plt.yticks([])
plt.show()
```
